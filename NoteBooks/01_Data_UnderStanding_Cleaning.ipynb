### Duplicates

```python
print("Duplicated Rows:", df.duplicated().sum())

print("Duplicated User IDs:", df['User_ID'].duplicated().sum())
```

```python
df[df.duplicated(keep=False)].sort_values('User_ID')
```

```python
# duplicated rows are exactly the same user, so i keep the first one only
df = df.drop_duplicates().reset_index(drop=True)
```

```python
print(df.shape)
print(df['User_ID'].duplicated().sum())
```

### Age

```python
df['Age'] = df['Age'].str.replace("yrs", "", regex=False).str.strip()
df['Age'] = pd.to_numeric(df['Age'])

# ages like 97 look like typos, so i will treat them as missing
print(df[df['Age'] > 80][['User_ID', 'Age', 'User_Type']])
```

```python
df.loc[df['Age'] > 80, 'Age'] = np.nan

# students and professionals have very different ages, so median per user type is better than one mean for all
df['Age'] = df['Age'].fillna(df.groupby('User_Type')['Age'].transform('median'))
df['Age'] = df['Age'].astype(int)

df['Age'].describe()
```

### Gender

```python
df['Gender'] = df['Gender'].str.strip().str.lower()
df['Gender'] = df['Gender'].replace({
    'm': "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
    "other": "Other",
    "non-binary": "Other"
})

# only the real missing values are Unknown (before i put "other" inside Unknown too, which is not the same thing)
df['Gender'] = df['Gender'].fillna("Unknown")

df['Gender'].value_counts(dropna=False)
```

### Country

```python
df['Country'] = df['Country'].str.strip().str.title().fillna("Unknown")
df['Country'] = df['Country'].replace({
    "U.K.": "United Kingdom",
    "Uk": "United Kingdom",
    "United States Of America": "United States",
    "Usa": "United States"
})

df['Country'].value_counts(dropna=False)
```

### Education

```python
df['Education_Level'] = df['Education_Level'].str.strip()
df['Education_Level'] = df['Education_Level'].replace({
    "Bachelor's" : "Bachelor's",
    "BA/BSc": "Bachelor's",
    "Bachelors": "Bachelor's",
    "bachelor's degree": "Bachelor's",
    "undergrad": "Undergraduate",
    "Masters": "Master's",
    "MA/MSc": "Master's",
    "master's degree": "Master's",
    "Ph.D.": "PhD",
    "high school": "High School",
    "HS": "High School"
})
df['Education_Level'] = df['Education_Level'].fillna("Unknown")

df['Education_Level'].value_counts(dropna=False)
```

```python
# some rows are strange, for example age 16 with a PhD
# i will not delete them (they are not many) but i add a flag so i can check them later
df['Age_Edu_Mismatch'] = (
    (df['Age'] < 18) & (df['Education_Level'].isin(["Master's", "PhD"]))
).astype(int)

print(df['Age_Edu_Mismatch'].sum(), "rows have age under 18 with Master's/PhD")
```

### Profession

```python
df['Profession'] = df['Profession'].str.strip().fillna("Unknown")

df['Profession'].nunique()
```

There are more than 120 different professions (some of them in Portuguese), too many to analyse.
So i group them in a few fields using keywords.

```python
prof = df['Profession'].str.lower()

conditions = [
    prof.str.contains("software|developer|desenvolved|programmer|computer|data|dados|machine learning|devops|web"),
    prof.str.contains("engineer|engineering"),
    prof.str.contains("research|scien|professor|teacher|lecturer|academic|literature|biology|physics|chemistry|math|education|history|psychology"),
    prof.str.contains("writer|content|journalist|design|multimedia|editor|creative|marketing|media"),
    prof.str.contains("support|customer|service"),
    prof.str.contains("business|analyst|finance|account|manager|sales|consult|admin|office"),
    prof.str.contains("medic|nurse|doctor|health|pharma"),
    prof.str.contains("unknown")
]

groups = [
    "Software & Data",
    "Engineering",
    "Science & Education",
    "Content & Media",
    "Customer Support",
    "Business & Admin",
    "Healthcare",
    "Unknown"
]

df['Profession_Group'] = np.select(conditions, groups, default="Other")

df['Profession_Group'].value_counts()
```

```python
# check what stayed in Other, if there are many i add more keywords above
df[df['Profession_Group'] == "Other"]['Profession'].value_counts()
```

### Monthly Income

```python
df['Monthly_Income'] = df['Monthly_Income'].str.replace(r'[$,]', '', regex=True)
df['Monthly_Income'] = pd.to_numeric(df['Monthly_Income'])

df.groupby('User_Type')['Monthly_Income'].describe()
```

```python
# a student with 250000 income is clearly wrong
# before i replaced it with 220 by hand, but that number is made up, so now i make it missing and fill it like the others
df.loc[(df['User_Type'] == 'Student') & (df['Monthly_Income'] > 5000), 'Monthly_Income'] = np.nan

# income depends on user type a lot (students vs professionals), so median per user type
df['Monthly_Income'] = df['Monthly_Income'].fillna(df.groupby('User_Type')['Monthly_Income'].transform('median'))
```

```python
print(df['Monthly_Income'].describe())
print(df['Monthly_Income'].isnull().sum())

# zero income is a real value here (mostly students?) so i keep it, but lets check who they are
print(df[df['Monthly_Income'] == 0]['User_Type'].value_counts())
```

### AI Tool & Purpose

```python
df['AI_Tool'] = df['AI_Tool'].str.strip()

df['AI_Tool'] = df['AI_Tool'].replace({
    "chatgpt": "ChatGPT",
    "Chat GPT": "ChatGPT",
    "Chatgpt": "ChatGPT",
    "Gemini": "Google Gemini",
    "gemini": "Google Gemini",
    "Copilot": "Microsoft Copilot",
    "copilot": "Microsoft Copilot",
    "claude": "Claude"
})

df['AI_Tool'] = df['AI_Tool'].fillna("Unknown")

df['AI_Tool'].value_counts(dropna=False)
```

```python
df['AI_Purpose'] = df['AI_Purpose'].fillna("Unknown")
df['AI_Purpose'].value_counts(dropna=False)
```

### AI Usage Hours

```python
df['AI_Usage_Hours_Per_Day'] = df['AI_Usage_Hours_Per_Day'].str.replace("hrs", "", regex=False).str.strip()
df['AI_Usage_Hours_Per_Day'] = pd.to_numeric(df['AI_Usage_Hours_Per_Day'])

print(df['AI_Usage_Hours_Per_Day'].sort_values(ascending=False).head(10).values)
```

```python
# 23 hours of AI per day is not possible, so anything above 16 hours is treated as wrong value
df.loc[df['AI_Usage_Hours_Per_Day'] > 16, 'AI_Usage_Hours_Per_Day'] = np.nan
df['AI_Usage_Hours_Per_Day'] = df['AI_Usage_Hours_Per_Day'].fillna(df['AI_Usage_Hours_Per_Day'].median())
```

```python
df['AI_Usage_Hours_Per_Day'].describe()
```

### Tasks, Cost and Scores

```python
df['Tasks_Performed'] = df['Tasks_Performed'].fillna("Unknown")
```

```python
df["Monthly_AI_Cost"] = df["Monthly_AI_Cost"].str.replace("$", "", regex=False)
df["Monthly_AI_Cost"] = pd.to_numeric(df["Monthly_AI_Cost"])

df['Monthly_AI_Cost'] = df['Monthly_AI_Cost'].fillna(df['Monthly_AI_Cost'].median())
df["Monthly_AI_Cost"].value_counts(dropna=False)
```

```python
df['Productivity_Score'] = df['Productivity_Score'].fillna(df['Productivity_Score'].median())
df["Productivity_Score"].describe()
```

```python
# accuracy rating is a 2 to 5 rating, median gave 3.5 which is not a real choice
# so i use the most common value (mode) instead
df["Accuracy_Rating"] = df["Accuracy_Rating"].fillna(df["Accuracy_Rating"].mode()[0])
df["Accuracy_Rating"].value_counts()
```

```python
df["Satisfaction_Score"] = df["Satisfaction_Score"].fillna(df["Satisfaction_Score"].median())
df["Satisfaction_Score"].value_counts(dropna=False)
```

```python
df["Time_Saved_Hours_Per_Week"] = df["Time_Saved_Hours_Per_Week"].str.replace("hrs", "", regex=False).str.strip()
df["Time_Saved_Hours_Per_Week"] = pd.to_numeric(df["Time_Saved_Hours_Per_Week"])

df["Time_Saved_Hours_Per_Week"] = df["Time_Saved_Hours_Per_Week"].fillna(df["Time_Saved_Hours_Per_Week"].median())
df["Time_Saved_Hours_Per_Week"].describe()
```

```python
df["Work_or_Study_Hours_Per_Day"] = df["Work_or_Study_Hours_Per_Day"].str.replace("hrs", "", regex=False).str.strip()
df["Work_or_Study_Hours_Per_Day"] = pd.to_numeric(df["Work_or_Study_Hours_Per_Day"])

print(df["Work_or_Study_Hours_Per_Day"].sort_values(ascending=False).head(10).values)
```

```python
# more than 16 hours of work/study per day is not realistic
df.loc[df["Work_or_Study_Hours_Per_Day"] > 16, "Work_or_Study_Hours_Per_Day"] = np.nan
df["Work_or_Study_Hours_Per_Day"] = df["Work_or_Study_Hours_Per_Day"].fillna(df["Work_or_Study_Hours_Per_Day"].median())
```

### Would Recommend

```python
# before i filled this with "Maybe" but a missing answer is not a "Maybe", so it is Unknown now
df['Would_Recommend'] = df['Would_Recommend'].fillna("Unknown")
df['Would_Recommend'].value_counts()
```

***Feature Engineering***
-----------------
New columns i will use in the EDA.

```python
# paid or free user
df['Plan'] = np.where(df['Monthly_AI_Cost'] > 0, 'Paid', 'Free')

# age groups
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 17, 25, 35, 50, 120],
    labels=['Under 18', '18-25', '26-35', '36-50', 'Over 50']
)

# how heavy is the usage
df['Usage_Level'] = pd.cut(
    df['AI_Usage_Hours_Per_Day'],
    bins=[0, 1.5, 3, 5, 16],
    labels=['Light', 'Moderate', 'Heavy', 'Very Heavy']
)

# hours saved for every hour of AI use (week hours / week hours)
df['Time_Saved_Ratio'] = df['Time_Saved_Hours_Per_Week'] / (df['AI_Usage_Hours_Per_Day'] * 7)
```
